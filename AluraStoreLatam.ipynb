### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Análisis de facturación total (todas las tiendas)

# Unimos los dataframes de las tiendas
df_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)

# Mostramos las primeras filas para revisar las columnas
df_tiendas.head()

# Calculamos la facturación total
facturacion_total = df_tiendas['precio'].sum()
print(f"Facturación total de todas las tiendas: ${facturacion_total:,.2f}")

# Si quieres la facturación por tienda individual:
facturacion_por_tienda = {
    'Tienda 1': tienda['precio'].sum(),
    'Tienda 2': tienda2['precio'].sum(),
    'Tienda 3': tienda3['precio'].sum(),
    'Tienda 4': tienda4['precio'].sum()
}
print("Facturación por tienda:")
for tienda_nombre, total in facturacion_por_tienda.items():
    print(f"{tienda_nombre}: ${total:,.2f}")
```

# 2. Ventas por categoría

```python
# Ventas por categoría

# Agrupamos por la columna de categoría y sumamos las ventas (precio)
ventas_por_categoria = df_tiendas.groupby('categoria')['precio'].sum().sort_values(ascending=False)

print("Ventas por categoría:")
print(ventas_por_categoria)
```

# 3. Calificación promedio de la tienda


```python
# Calificación promedio de todas las tiendas
calificacion_promedio = df_tiendas['calificacion'].mean()
print(f"Calificación promedio de todas las tiendas: {calificacion_promedio:.2f}")

# Calificación promedio por tienda individual
calificacion_por_tienda = {
    'Tienda 1': tienda['calificacion'].mean(),
    'Tienda 2': tienda2['calificacion'].mean(),
    'Tienda 3': tienda3['calificacion'].mean(),
    'Tienda 4': tienda4['calificacion'].mean()
}
print("Calificación promedio por tienda:")
for tienda_nombre, promedio in calificacion_por_tienda.items():
    print(f"{tienda_nombre}: {promedio:.2f}")
```

# 4. Productos más y menos vendidos

```python
# Productos más y menos vendidos

# Agrupamos por producto y sumamos la cantidad vendida
ventas_por_producto = df_tiendas.groupby('producto')['cantidad'].sum().sort_values(ascending=False)

print("Top 5 productos más vendidos:")
print(ventas_por_producto.head(5))

print("\nTop 5 productos menos vendidos:")
print(ventas_por_producto.tail(5))
```
